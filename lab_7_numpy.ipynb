# **Lab 7 - Numpy**

## Task 1 - Load and Extract Stock Data 

```python
import numpy as np 
```

```python
def load_extract_stock_data(filename = "stock_prices_intro.csv"):

    data = np.genfromtxt(filename, delimiter= ",", names= True, dtype= None, encoding= "utf-8")

    date =  data["Date"]
    openp = data["Open"]
    closep = data["Close"]
    vol = data["Volume"]

    print("Date:", date)
    print("Opening Price:", openp)
    print("Closing Price:", closep)
    print("Volume:", vol)

    return date, openp, closep, vol
```

```python
load_extract_stock_data()
```

## Task 2 - Calculate Daily Returns 

```python
def calculate_daily_returns(openp, closep):

    rate = ((closep - openp) / openp) * 100

    print(f"Rate: {np.round(rate[:5], 2)}%")

    return rate
```

```python
date, openp, closep, vol = load_extract_stock_data()
```

```python
rate = calculate_daily_returns(openp, closep)
```

## Task 3 - Calculate Basic Statistics

```python
def calculate_basic_statistics(rate):

    # Calculate Basic Statistics 

    mean = np.mean(rate)
    median = np.median(rate)
    standard_deviation = np.std(rate)

    # Display Statistics 
    print(f"Mean: {mean:.2f}%")
    print(f"Median: {median:.2f}%")
    print(f"Standard Deviation: {standard_deviation:.2f}%")

    return mean, median, standard_deviation
```

```python
calculate_basic_statistics(rate)
```

## Task 4 - Identify the Best and Worst Trading Days 

```python
def best_worst_trading_days(rate, openp, closep, date):
    
    
    best_day_index = np.argmax(rate)
    worst_day_index = np.argmin(rate)

    print("Best Trading Day:")
    print(f"Date: {date[best_day_index]}")
    print(f"Opening Price: {openp[best_day_index]}")
    print(f"Closing Price: {closep[best_day_index]}")
    print(f"Daily Return Rate: {rate[best_day_index]:.2f}%")

    print("\nWorst Trading Day:")
    print(f"Date: {date[worst_day_index]}")
    print(f"Opening Price: {openp[worst_day_index]}")
    print(f"Closing Price: {closep[worst_day_index]}")
    print(f"Daily Return Rate: {rate[worst_day_index]:.2f}%")

    return best_day_index, worst_day_index
```

```python
best_worst_trading_days(rate, openp, closep, date)
```

## Task 5 - Analyze Up and Down Trading Days

```python
def analyze_up_down_days(rate):
    positive_return_days = np.sum(rate > 0)
    negative_return_days = np.sum(rate < 0)

    print(f"Positive Return Days: {positive_return_days} Days")
    print(f"Negative Return Days: {negative_return_days} Days")

    return positive_return_days, negative_return_days
```

```python
analyze_up_down_days(rate)
```

## Task 6 - Calculate Cumulative Return 

```python
def calculate_cumulative_return(rate):
    f = 1 + rate / 100
    cum_return = (np.prod(f) - 1) * 100
    print(f"Cumulative Return: {cum_return:.2f}%")

    return cum_return
```

```python
calculate_cumulative_return(rate)
```

## Normal Case

## Normal Case – Explanation

### The normal case tests the program using the original stock dataset containing 20 business days. It verifies that the CSV file loads correctly, the four columns are extracted, and all six tasks produce the expected results.

```python
# NORMAL CASE: Test the actual stock dataset

# Test 1: Load the stock data
date, openp, closep, vol = load_extract_stock_data()

expected_records = 20
actual_records = len(date)

print("Test 1 - Number of Records")
print("Expected:", expected_records)
print("Actual:", actual_records)
print("Status:", "PASSED" if actual_records == expected_records else "FAILED")


# Test 2: Verify the extracted arrays
arrays_correct = (
    len(date) == len(openp) == len(closep) == len(vol) == 20
)

print("\nTest 2 - Extracted Arrays")
print("Expected: Four arrays containing 20 records each")
print("Actual:", len(date), len(openp), len(closep), len(vol))
print("Status:", "PASSED" if arrays_correct else "FAILED")


# Test 3: Verify daily return calculations
rate = calculate_daily_returns(openp, closep)

expected_first_five = np.array([2.03, -1.31, -0.19, 1.32, -0.76])

actual_first_five = np.round(rate[:5], 2)

print("\nTest 3 - First Five Daily Returns")
print("Expected:", expected_first_five)
print("Actual:", actual_first_five)
print("Status:", "PASSED" if np.allclose(actual_first_five, expected_first_five) else "FAILED")


# Test 4: Verify basic statistics
mean, median, sd = calculate_basic_statistics(rate)

print("\nTest 4 - Basic Statistics")
print("Expected: Mean=0.12, Median=0.18, SD=1.18")
print(f"Actual: Mean={mean:.2f}, Median={median:.2f}, SD={sd:.2f}")

statistics_correct = (
    np.isclose(mean, 0.12, atol=0.01) and
    np.isclose(median, 0.18, atol=0.01) and
    np.isclose(sd, 1.18, atol=0.01)
)

print("Status:", "PASSED" if statistics_correct else "FAILED")


# Test 5: Verify best and worst trading days
best, worst = best_worst_trading_days(rate, openp, closep, date)

print("\nTest 5 - Best and Worst Trading Days")
print("Expected Best:", "2025-09-26")
print("Actual Best:", date[best])
print("Expected Worst:", "2025-09-22")
print("Actual Worst:", date[worst])

days_correct = (
    date[best] == "2025-09-26" and
    date[worst] == "2025-09-22"
)

print("Status:", "PASSED" if days_correct else "FAILED")


# Test 6: Verify positive and negative trading days
positive, negative = analyze_up_down_days(rate)

print("\nTest 6 - Positive and Negative Days")
print("Expected: Positive=11, Negative=9")
print(f"Actual: Positive={positive}, Negative={negative}")

counts_correct = positive == 11 and negative == 9

print("Status:", "PASSED" if counts_correct else "FAILED")


# Test 7: Verify cumulative return
cum_return = calculate_cumulative_return(rate)

print("\nTest 7 - Cumulative Return")
print("Expected: 2.33%")
print(f"Actual: {cum_return:.2f}%")

print("Status:", "PASSED" if np.isclose(cum_return, 2.33, atol=0.01) else "FAILED")
```

## Special Case

## Special Case – Explanation

### Special cases test unusual but valid situations that the program should handle correctly. These include zero daily returns, tied maximum returns, and datasets where every trading day has a negative return.

```python
# SPECIAL CASE 1: All trading days have zero returns

special_date = np.array(["Day 1", "Day 2", "Day 3", "Day 4"])
special_open = np.array([100, 100, 100, 100])
special_close = np.array([100, 100, 100, 100])

special_rate = calculate_daily_returns(special_open, special_close)

print("\nSpecial Case 1 - Zero Returns")
print("Expected:", [0, 0, 0, 0])
print("Actual:", special_rate)
print("Status:", "PASSED" if np.allclose(special_rate, [0, 0, 0, 0]) else "FAILED")

mean, median, sd = calculate_basic_statistics(special_rate)

positive, negative = analyze_up_down_days(special_rate)
cum_return = calculate_cumulative_return(special_rate)

print("Expected counts: Positive=0, Negative=0")
print(f"Actual counts: Positive={positive}, Negative={negative}")
print("Expected cumulative return: 0.00%")
print(f"Actual cumulative return: {cum_return:.2f}%")


# SPECIAL CASE 2: Multiple trading days share the highest return

special_date = np.array(["Day 1", "Day 2", "Day 3", "Day 4"])
special_open = np.array([100, 100, 100, 100])
special_close = np.array([105, 98, 105, 101])

special_rate = calculate_daily_returns(special_open, special_close)

best, worst = best_worst_trading_days(
    special_rate, special_open, special_close, special_date
)

print("\nSpecial Case 2 - Tied Highest Returns")
print("Expected best index: 0 (first occurrence)")
print("Actual best index:", best)
print("Expected worst index: 1")
print("Actual worst index:", worst)

print("Status:", "PASSED" if best == 0 and worst == 1 else "FAILED")


# SPECIAL CASE 3: All trading days have negative returns

special_date = np.array(["Day 1", "Day 2", "Day 3", "Day 4"])
special_open = np.array([100, 100, 100, 100])
special_close = np.array([98, 95, 99, 97])

special_rate = calculate_daily_returns(special_open, special_close)

positive, negative = analyze_up_down_days(special_rate)

best, worst = best_worst_trading_days(
    special_rate, special_open, special_close, special_date
)

print("\nSpecial Case 3 - All Negative Returns")
print("Expected positive days: 0")
print("Actual positive days:", positive)
print("Expected negative days: 4")
print("Actual negative days:", negative)
print("Expected best index: 2")
print("Actual best index:", best)
print("Expected worst index: 1")
print("Actual worst index:", worst)

print("Status:", "PASSED" if (
    positive == 0 and negative == 4 and best == 2 and worst == 1
) else "FAILED")
```
